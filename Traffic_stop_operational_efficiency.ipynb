```python
# Data audit for numerical data

def continuous_variable_summary(x):
    #freq and missing values
    n_total=x.shape[0]
    n_miss=x.isna().sum()
    perc_miss=n_miss/n_total*100

    #Outliers
    q1=x.quantile(0.25)
    q3=x.quantile(0.75)
    iqr=q3-q1
    lc_iqr=q1-1.5*iqr
    uc_iqr=q3+1.5*iqr
    return pd.Series([x.dtype,x.nunique(),n_total,n_miss,perc_miss,x.sum(),x.mean(),x.std(),x.var(),lc_iqr,uc_iqr,x.min(),
                      x.quantile(0.01),x.quantile(0.05),x.quantile(0.10),x.quantile(0.25),x.median(),
                      x.quantile(0.75),x.quantile(0.90),x.quantile(0.95),x.quantile(0.99),x.max()],index=
                     ['dtype','cardinality','n_tot','miss','perc_miss','sum','mean','std','var','lc_iqr','uc_iqr'
                      ,'min','p1','p5','p10','p25','p50','p75','p90','p95','p99','max'])
```

```python
police.select_dtypes(['float64','int64']).apply(continuous_variable_summary)
```

```python
# Data audit for categorical data

# EDA categorical data
def categorical_var_summary(x):
    Mode = x.value_counts().sort_values(ascending = False)[0:1].reset_index()
    return pd.Series([x.count(), x.isnull().sum(), Mode.iloc[0, 0], Mode.iloc[0, 1], 
                          round(Mode.iloc[0, 1] * 100/x.count(), 2),x.nunique()], 
                  index = ['N', 'NMISS', 'MODE', 'FREQ', 'PERCENT','CARDINALITY'])
```

```python
police.select_dtypes(['str','object']).apply(categorical_var_summary)
```

```python
# Outlier treatment

# Outlier treatment on numerical column

def outlier(x):
    q1=x.quantile(0.25)
    q3=x.quantile(0.75)
    iqr=q3-q1
    lc_iqr=q1-1.5*iqr
    uc_iqr=q3+1.5*iqr
    return x.clip(lower=lc_iqr,upper=uc_iqr)
```

```python
num_cols=police.select_dtypes(['int64','float64']).columns
police[num_cols]=police[num_cols].apply(outlier)
```

```python
# Missing value imputation

def missing_imputation(x):
    if (x.dtype=='float64') | (x.dtype=='int64'):
        x=x.fillna(x.mean())
    elif (x.dtype=='str') | (x.dtype=='object'):
        x=x.fillna(x.mode()[0])
    return(x)
```

```python
police=police.apply(missing_imputation)
```
